```python
import os, shutil
import numpy as np
import h5py
```

```python
# ── Parameters ───────────────────────────────────────────────────────────────
INPUT         = '~/../../scratch/gpfs/MELCHIOR/cj1223/DIB/'
INPUT +=  "bc_run_SymmetryC2v_BCTrue_F1p0_DTrue_FlatFalse_SpecTrue_dTTrue_covFalse_nonlinFalse_tauSlope0p05_update.h5"
# Pairs of indices to swap: [(i, j), ...]. Empty list = no swaps.
SWAP_PAIRS    = []   # e.g. [(1, 3)] or [(1, 3), (4, 5)]

# Set to True to append a new dimension sampled from N(NEW_DIM_LOC, NEW_DIM_SCALE)
ADD_DIM       = True
NEW_DIM_LOC   = 0.0
NEW_DIM_SCALE = 0.01

SEED          = None   # set to an int for reproducibility
```

```python
# ── Run ──────────────────────────────────────────────────────────────────────
assert SWAP_PAIRS or ADD_DIM, 'Nothing to do: set SWAP_PAIRS and/or ADD_DIM'

src = os.path.abspath(os.path.expanduser(INPUT))
assert os.path.exists(src), f'Input file not found: {src}'

base, ext = os.path.splitext(src)
dst = base + '_prep_restart' + ext
if os.path.exists(dst):
    os.remove(dst)
    print(f'Overwriting existing output.')

print(f'Input:  {src}')
print(f'Output: {dst}')

shutil.copy2(src, dst)
print('Copied.')

if SEED is not None:
    np.random.seed(SEED)

with h5py.File(dst, 'r+') as f:
    chain = f['mcmc/chain'][:]           # (nsteps, nwalkers, ndim)
    nsteps, nwalkers, ndim = chain.shape
    print(f'Chain shape: {chain.shape}')

    for i, j in SWAP_PAIRS:
        assert 0 <= i < ndim and 0 <= j < ndim, \
            f'Swap index out of range: ({i}, {j}) — ndim={ndim}'
        chain[:, :, [i, j]] = chain[:, :, [j, i]]
        print(f'  Swapped dim {i} ↔ dim {j}')

    if ADD_DIM:
        new_col = np.random.normal(loc=NEW_DIM_LOC, scale=NEW_DIM_SCALE,
                                   size=(nsteps, nwalkers, 1))
        chain = np.concatenate([chain, new_col], axis=2)
        print(f'  Appended dim {ndim} ~ N(loc={NEW_DIM_LOC}, scale={NEW_DIM_SCALE})')
        print(f'  New chain shape: {chain.shape}')

    del f['mcmc/chain']
    f.create_dataset('mcmc/chain', data=chain)

print(f'Done → {os.path.basename(dst)}')
```
